```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
import os

# physics constants (using lost city meteorite specs)
const = {
    'g': 9.81,  # gravity (m/s^2)
    'rho_a': 1.225,  # sea-level air density (kg/m^3)
    'Cd': 0.47,  # drag coefficient (assuming sphere)
    'rho_p': 3700,  # particle density (kg/m^3)
    'H': 8500.0,  # atmospheric scale height (m)
    'K_abl': 1.25e-8  # ablation coefficient (s^2/m^2)
}


# core physics engine for atmospheric entry
def atm_entry(t, st):
    x, y, vx, vy, m = st

    # bail out early if the particle burned up completely or stopped falling
    if m <= 1e-15 or abs(vy) < 1e-6 or y <= 1e-15:
        return [0, 0, 0, 0, 0]

    g = const['g']
    rho_p = const['rho_p']
    rho_a = const['rho_a']
    Cd = const['Cd']
    H = const['H']
    K_abl = const['K_abl']

    # local atmospheric density
    rho = rho_a * np.exp(-y / H)

    # watch out for negative mass which breaks the radius calculation
    if m <= 0:
        r = 0
    else:
        r = (3 * m / (4 * np.pi * rho_p)) ** (1 / 3)

    A = np.pi * (r ** 2)
    v = np.sqrt(vx * vx + vy * vy)

    # avoid division by zero if it's basically gone
    if m < 1e-15:
        return [0, 0, 0, 0, 0]

    # calculate the derivatives
    dxdt = vx
    dydt = vy
    dvxdt = -rho * Cd * A * v * vx / (2 * m)
    dvydt = -g - rho * Cd * A * v * vy / (2 * m)
    dmdt = -K_abl * rho * A * (v ** 3) / 2

    return [dxdt, dydt, dvxdt, dvydt, dmdt]


# event hooks to tell the solver when to stop early

def hit_ground(t, st):
    # stop the sim when it hits the dirt
    return st[1]


hit_ground.terminal = True
hit_ground.direction = -1


def burn_up(t, st):
    # stop if mass is negligible
    return st[4] - 1e-15


burn_up.terminal = True

# default starting conditions
y0 = 100 * 10 ** 3  # start at the karman line (100km)
x0 = 0

# max simulation time
t_span = (0, 50)
t_eval = np.linspace(*t_span, 500)


def run_simulation(r0_microns, v0_kms, angle_deg):
    # prep units and initial state
    r0 = r0_microns * 1e-6
    m0 = (r0 ** 3) * 4 * np.pi * const['rho_p'] / 3

    v0 = v0_kms * 1000
    rad_theta = angle_deg * np.pi / 180

    vx0 = v0 * np.cos(rad_theta)
    vy0 = -v0 * np.sin(rad_theta)

    st = [x0, y0, vx0, vy0, m0]

    # run the math
    sol = solve_ivp(
        fun=atm_entry,
        t_span=t_span,
        y0=st,
        t_eval=t_eval,
        events=[hit_ground, burn_up],
        method='RK45'
    )

    # process the output for plotting
    v_total_km_s = np.sqrt(sol.y[2] ** 2 + sol.y[3] ** 2) / 1000
    alt_km = sol.y[1] / 1000
    mass_kg = sol.y[4]

    # calc radius safely, handling potential negative mass from numerical noise
    r_micr = np.cbrt(np.maximum(0, 3 * mass_kg / (4 * np.pi * const['rho_p']))) * 1e6

    # format legend label
    label = f"r={r0_microns:.0f}μm, v={v0_kms}km/s, θ={angle_deg}°"

    return {
        't': sol.t,
        'alt_km': alt_km,
        'v_total_km_s': v_total_km_s,
        'r_micr': r_micr,
        'label': label
    }


def plot_simulation_results(results_list, main_title, filename_prefix):
    # 1. height vs time
    plt.figure(figsize=(8, 5))
    for res in results_list:
        plt.plot(res['t'], res['alt_km'], label=res['label'])
    plt.xlabel('Time (s)')
    plt.ylabel('Altitude (km)')
    plt.title(f"{main_title}\nAltitude vs. Time")
    plt.grid(True)
    plt.legend(fontsize='small')
    filename_1 = f"{filename_prefix}_1_alt_vs_time.png"
    plt.savefig(filename_1)
    print(f"saved plot to {os.path.abspath(filename_1)}")
    plt.close()

    # 2. total velocity vs time
    plt.figure(figsize=(8, 5))
    for res in results_list:
        plt.plot(res['t'], res['v_total_km_s'], label=res['label'])
    plt.xlabel('Time (s)')
    plt.ylabel('Total Velocity (km/s)')
    plt.title(f"{main_title}\nTotal Velocity vs. Time")
    plt.grid(True)
    plt.legend(fontsize='small')
    filename_2 = f"{filename_prefix}_2_vel_vs_time.png"
    plt.savefig(filename_2)
    print(f"saved plot to {os.path.abspath(filename_2)}")
    plt.close()

    # 3. radius vs time
    plt.figure(figsize=(8, 5))
    for res in results_list:
        plt.plot(res['t'], res['r_micr'], label=res['label'])
    plt.xlabel('Time (s)')
    plt.ylabel('Radius (μm)')
    plt.title(f"{main_title}\nRadius vs. Time")
    plt.grid(True)
    plt.legend(fontsize='small')
    plt.ylim(bottom=0)  # keep radius visually clamped at 0
    filename_3 = f"{filename_prefix}_3_radius_vs_time.png"
    plt.savefig(filename_3)
    print(f"saved plot to {os.path.abspath(filename_3)}")
    plt.close()

    # 4. velocity vs altitude
    plt.figure(figsize=(8, 5))
    for res in results_list:
        plt.plot(res['alt_km'], res['v_total_km_s'], label=res['label'])
    plt.xlabel('Altitude (km)')
    plt.ylabel('Velocity (km/s)')
    plt.title(f"{main_title}\nVelocity vs. Altitude")
    plt.grid(True)
    plt.legend(fontsize='small')
    plt.xlim(left=40, right=y0 / 1000)
    filename_4 = f"{filename_prefix}_4_vel_vs_alt.png"
    plt.savefig(filename_4)
    print(f"saved plot to {os.path.abspath(filename_4)}")
    plt.close()

    # 5. radius vs altitude
    plt.figure(figsize=(8, 5))
    for res in results_list:
        plt.plot(res['alt_km'], res['r_micr'], label=res['label'])
    plt.xlabel('Altitude (km)')
    plt.ylabel('Radius (μm)')
    plt.title(f"{main_title}\nRadius vs. Altitude")
    plt.grid(True)
    plt.legend(fontsize='small')
    plt.xlim(left=0, right=y0 / 1000)
    plt.ylim(bottom=0)
    filename_5 = f"{filename_prefix}_5_radius_vs_alt.png"
    plt.savefig(filename_5)
    print(f"saved plot to {os.path.abspath(filename_5)}")
    plt.close()


# main execution block
if __name__ == "__main__":

    # case 1: keep speed & size constant, test different angles
    print("generating plot set 1: testing entry angles...")
    v0_const_1 = 15
    r0_const_1 = 100
    angles_to_test = [30, 45, 60, 90]

    results_case_1 = []
    for angle in angles_to_test:
        sim_params = {'r0_microns': r0_const_1, 'v0_kms': v0_const_1, 'angle_deg': angle}
        results_case_1.append(run_simulation(**sim_params))

    plot_simulation_results(
        results_case_1,
        main_title=f"Effect of Angle (v0={v0_const_1} km/s, r0={r0_const_1} μm)",
        filename_prefix="1_plot_different_angles"
    )

    # case 2: keep speed & angle constant, test different particle sizes
    print("generating plot set 2: testing particle sizes...")
    v0_const_2 = 20
    angle_const_2 = 45
    radii_to_test = [5, 10, 50, 100]

    results_case_2 = []
    for r0 in radii_to_test:
        sim_params = {'r0_microns': r0, 'v0_kms': v0_const_2, 'angle_deg': angle_const_2}
        results_case_2.append(run_simulation(**sim_params))

    plot_simulation_results(
        results_case_2,
        main_title=f"Effect of Size (v0={v0_const_2} km/s, θ={angle_const_2}°)",
        filename_prefix="2_plot_different_sizes"
    )

    # case 3: keep size & angle constant, test different entry speeds
    print("generating plot set 3: testing initial speeds...")
    r0_const_3 = 100
    angle_const_3 = 45
    speeds_to_test = [12, 15, 20, 25]

    results_case_3 = []
    for v0 in speeds_to_test:
        sim_params = {'r0_microns': r0_const_3, 'v0_kms': v0, 'angle_deg': angle_const_3}
        results_case_3.append(run_simulation(**sim_params))

    plot_simulation_results(
        results_case_3,
        main_title=f"Effect of Speed (r0={r0_const_3} μm, θ={angle_const_3}°)",
        filename_prefix="3_plot_different_speeds"
    )

    print("all done. plots generated successfully.")
```

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
import time
from scipy.interpolate import RegularGridInterpolator
import matplotlib.colors as mcolors

print("initializing script...")

# physical constants using the lost city meteorite as a reference
const = {
    'g': 9.81,  # gravity (m/s^2)
    'rho_a': 1.225,  # sea-level air density (kg/m^3)
    'Cd': 0.47,  # drag coefficient for a rough sphere
    'rho_p': 3700,  # particle density (kg/m^3)
    'H': 8500.0,  # atmospheric scale height (m)
    'K_abl': 6.25e-9  # ablation coefficient (s^2/m^2)
}

# core physics equations for the solver
def atm_entry(t, st):
    # unpack state. st[4] tracks the shrinking radius
    x, y, vx, vy, r = st

    # kill the simulation if it burned up, stopped moving vertically, or hit the ground
    if r <= 1e-7 or abs(vy) < 1e-6 or y <= 1e-15:
        return [0, 0, 0, 0, 0]

    g = const['g']
    rho_p = const['rho_p']
    rho_a = const['rho_a']
    Cd = const['Cd']
    H = const['H']
    K_abl = const['K_abl']

    # calc current mass and area from the radius
    m = (4.0 / 3.0) * np.pi * (r ** 3) * rho_p
    A = np.pi * (r ** 2)

    # local atmospheric density and total velocity
    rho = rho_a * np.exp(-y / H)
    v = np.sqrt(vx * vx + vy * vy)

    # derivatives for position and velocity
    dxdt = vx
    dydt = vy
    dvxdt = -rho * Cd * A * v * vx / (2 * m)
    dvydt = -g - rho * Cd * A * v * vy / (2 * m)

    # radius derivative (how fast it's ablating/melting)
    drdt = -(K_abl * rho * (v ** 3)) / (8 * rho_p)

    return [dxdt, dydt, dvxdt, dvydt, drdt]


# event hooks for the integrator to know when to terminate early

def hit_ground(t, st):
    return st[1]

hit_ground.terminal = True
hit_ground.direction = -1

def burn_up(t, st):
    return st[4] - 1e-7

burn_up.terminal = True

# baseline starting parameters
x0 = 0
y0 = 100 * 10 ** 3  # drop from the karman line (100km)
t_span = (0, 50)  # max 50 seconds of flight time

print("setup complete. configuring parameter sweep...")

# locking entry angle to 45 degrees for this run
fixed_deg_theta = 45
rad_theta = fixed_deg_theta * np.pi / 180

# grid resolution for the parameter sweep
N_v = 30
N_r = 30

# sweeping across entry speeds and initial sizes
v0_kms_list_map = np.linspace(10, 25, N_v)
r0_um_list_map = np.linspace(10, 200, N_r)

# pre-allocate results matrix
final_radius_grid = np.zeros((N_r, N_v))

print(f"running {N_v * N_r} simulations...")
start_time = time.time()

# main loop to test every size/speed combo
for i, r0_um in enumerate(r0_um_list_map):
    for j, v0_kms in enumerate(v0_kms_list_map):

        # convert to standard SI units for the math
        r0 = r0_um * 1e-6
        v0 = v0_kms * 1000

        # split velocity into components
        vx0 = v0 * np.cos(rad_theta)
        vy0 = -v0 * np.sin(rad_theta)

        st = [x0, y0, vx0, vy0, r0]

        sol = solve_ivp(
            fun=atm_entry, t_span=t_span, y0=st,
            events=[hit_ground, burn_up], method='RK45'
        )

        # grab the final radius from the end of the simulation array
        final_radius_um = sol.y[4][-1] * 1e6

        # save it if it survived, otherwise log it as a 0
        if final_radius_um > 0.1:
            final_radius_grid[i, j] = final_radius_um
        else:
            final_radius_grid[i, j] = 0

print(f"simulations finished in {time.time() - start_time:.2f} seconds")
print("generating plots...")

# plot 1: the survival heatmap
fig1 = plt.figure(figsize=(10, 8))
ax = fig1.add_subplot(111)

# custom colormap so completely burned particles show up as black
cmap_radius = mcolors.LinearSegmentedColormap.from_list("mycmap", ["#440154", "#21908d", "#fde725"])
cmap_radius.set_under('black')

pcm = ax.pcolormesh(v0_kms_list_map, r0_um_list_map, final_radius_grid,
                    shading='auto', cmap='viridis', vmin=0.1)

cbar = plt.colorbar(pcm, label='Final Radius on Ground (µm)')
ax.set_xlabel('Initial Velocity, v0 (km/s)', fontsize=12)
ax.set_ylabel('Initial Radius, r0 (µm)', fontsize=12)
ax.set_title(f'Particle Survival Map\n(Entry Angle = {fixed_deg_theta}°)', fontsize=14)
ax.grid(True, linestyle=':', alpha=0.5)
ax.tick_params(axis='both', which='major', labelsize=10)
fig1.tight_layout()

# plot 2: before/after histograms showing the atmospheric filter effect
print("calculating distribution distortion...")

# build an interpolator from our grid data so we don't have to recalculate the heavy physics
filter_interpolator = RegularGridInterpolator(
    (r0_um_list_map, v0_kms_list_map),
    final_radius_grid,
    bounds_error=False, fill_value=0
)

N_particles = 100000

# mock up a random cloud of incoming particles
hypothetical_r0 = np.random.uniform(min(r0_um_list_map), max(r0_um_list_map), N_particles)
hypothetical_v0 = np.random.uniform(min(v0_kms_list_map), max(v0_kms_list_map), N_particles)

points_to_filter = np.column_stack((hypothetical_r0, hypothetical_v0))
final_particles_rf = filter_interpolator(points_to_filter)

fig2, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6), sharey=True)
fig2.suptitle('Atmospheric Filter Effect on Particle Distribution', fontsize=16)

# left side: the initial random distribution
ax1.hist(hypothetical_r0, bins=50, color='skyblue', edgecolor='black', alpha=0.8)
ax1.set_title('BEFORE (Deep Space)', fontsize=12)
ax1.set_xlabel('Initial Radius, r0 (µm)', fontsize=10)
ax1.set_ylabel('Particle Count', fontsize=10)
ax1.grid(True, linestyle=':', alpha=0.7)
ax1.tick_params(axis='both', which='major', labelsize=9)

# right side: what survived to hit the ground
survivors = final_particles_rf[final_particles_rf > 0.1]
ax2.hist(survivors, bins=50, color='salmon', edgecolor='black', alpha=0.8)
ax2.set_title('AFTER (Earth Surface)', fontsize=12)
ax2.set_xlabel('Final Radius, r_final (µm)', fontsize=10)
ax2.grid(True, linestyle=':', alpha=0.7)
ax2.tick_params(axis='both', which='major', labelsize=9)
fig2.tight_layout(rect=[0, 0.03, 1, 0.93])

print("\ndone. displaying plots.")
plt.show()

# dump some quick stats to the console
print("\nrun summary:")
print(f"total particles generated: {N_particles}")
print(f"survivors: {len(survivors)} ({len(survivors) / N_particles * 100:.1f}%)")
print(f"burned up: {N_particles - len(survivors)}")
```
